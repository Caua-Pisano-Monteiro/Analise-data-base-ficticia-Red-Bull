# Verificando se há dados do faturamento inconsistente

```python
teste_faturamento = dados['preco_unitario_brl'] * dados['quantidade_unidades']
teste_faturamento.to_frame(name='Valores') # Nomeando a coluna
```

teste_faturamento faz a operação de multiplicar o valor unitário pela quantidade, assim podendo comparar com a coluna faturamento_brl

```python
teste = (dados['faturamento_brl'] == teste_faturamento).all()
if teste:
    print('Valores iguais')
else:
    print('Contêm divergência')
```

Tendo o resulado final como 'Contêm divergência', vou verificar qual os índices que os valores não batem

```python
teste1 = dados[dados['faturamento_brl'] != teste_faturamento].index
teste1
```

```python
dados.iloc[7]
```

A verficando se o indice 7 está errado mesmo, tirando a conlusão que não. Se fizer a multiplicação de quantidade_unidade = 3 com preco_unitario_brl = 9.9,
temos o resultado igual o do faturamento_brl

Meu erro aqui foi não igualar o números de dígitos depois da vírgula

```python
teste = (dados['faturamento_brl'].round(2) == teste_faturamento.round(2)).all()
if teste:
    print('Valores iguais')
else:
    print('Contêm divergência')
```

Comparando se os valores de dados['faturamento_brl'] e teste_faturamento são iguais, sendo iguais, os dados['faturaamento_brl'] estão corretos sem inconsistência

Analisando o perfil de consumo com base na faixa etaria de idade

```python
consumidores_idade = (
    dados.groupby("faixa_etaria", observed=True)
    .agg(
        unidades_vendidas=("quantidade_unidades", "sum"),
        faturamento=("faturamento_brl", "sum"),
        consumidores=("id_consumidor", "nunique")
    )
    .sort_values("unidades_vendidas", ascending=False)
)

consumidores_idade
```

Analisando o consumo conforme a ocasiões

```python
ocasioes = (
    dados.groupby(
        ["faixa_etaria", "ocasiao_consumo"],
        observed=True
    )
    .agg(
        unidades=("quantidade_unidades", "sum"),
        faturamento=("faturamento_brl", "sum")
    )
    .reset_index() # Retirando os index 
    # Ordenando em que ordem aparacerar, sendo "faixa_etaria" do menor para o maior e "unidades" do maior para o maior.
    .sort_values(
        ["faixa_etaria", "unidades"],
        ascending=[True, False] 
    )# ascending - True(a-z) e False(z-a)
)

ocasioes
```
